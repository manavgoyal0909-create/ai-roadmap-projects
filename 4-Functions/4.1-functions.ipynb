## Functions in python

```python
## Syntax
def function_name(parameters):
    """Docstring"""
    # Function body
    return expression
```

```python
## Why functions?
num=23
if num%2==0:
    print("The number is even")
else:
    print("The numebr is odd")    
```

```python
def even_or_odd(num):
    """This function finds even or odd"""
    if num%2==0:
        print("The number is even")
    else:
        print("The numebr is odd")  
```

```python
## Call this function
even_or_odd(55555555553)
```

```python
## Function with multiple parameters

def add(a,b):
    return a+b

result=add(2,4)
print(result)
```

```python
## Default parameters

def greet(name="Guest"):
    print(f"Hello {name} Welcome to the paradise")
greet("Manav")    
```

```python
## Variable length arguments
## positional and keywords arguments

def print_numbers(*Manav):
    for number in Manav:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,"SUII")
```

```python
## Positional arguments
def print_numbers(*args):
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,"SUII")
```

```python
## keywords arguments

def print_detail(**kwargs):
    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_detail(name="Manav",age="15",country="India")
```

```python
def print_detail(*args,**kwargs):
    for val in args:
        print(f" Positional argument:{val}")

    for key,value in kwargs.items():
        print(f"{key}:{value}")
```

```python
print_detail(1,2,3,4,"SUII",name="Manav",age="15",country="India")
```

```python
## Return statement
def multiply(a,b):
    return a*b

multiply(2,3)
```

```python
## Return multiple parameters
def multiply(a,b):
    return a*b,a

multiply(2,3)
```
