## Lambda functions in python

```python
## Syntax
lambda arguments: expression
```

```python
def addition(a,b):
    return a+b
```

```python
addition(9,39)
```

```python
addition=lambda a,b:a+b
print(type(addition))
print(addition(3,23))
```

```python
def even(num):
    if num%2==0:
        return True


even(80)
```

```python
even1= lambda num:num%2==0
even1(46)
```

```python
def addition(x,y,z):
    return x+y+z

addition(34,2,4)    
```

```python
addition1= lambda x,y,z:x+y+z
addition1(23,2,3)
```

```python
## map()- applies a function to all items in a list

numbers=[1,2,3,4,5,6]
def square(number):
    return number**2

square(3)    
```

```python
list(map(lambda x:x**2,numbers))
```
